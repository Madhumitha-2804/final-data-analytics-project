### handle missing values

```python
df['children']=df['children'].fillna(0)

df['country']=df['country'].fillna('Unknown')

df['agent']=df['agent'].fillna(0)

df['company']=df['company'].fillna(0)
```

### Remove duplicates

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

### Correct data types

```python
numeric_columns = [
    "adults",
    "children",
    "babies",
    "is_canceled",
    "lead_time",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adr"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

### Remove unnecessary columns

```python
df = df.drop(columns=["agent","company"], errors="ignore")
```

### Total_Guests

```python
df["Total_Guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)
```

```python
df[["adults", "children", "babies", "Total_Guests"]].head()
```

### Stay_Nights

```python
df["Stay_Nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)
```

```python
df[
    [
        "stays_in_weekend_nights",
        "stays_in_week_nights",
        "Stay_Nights"
    ]
].head()
```

### Booking_Status

is_canceled = 1 → Cancelled
is_canceled = 0 → Completed

```python
df["Booking_Status"] = df["is_canceled"].map({
    1: "Cancelled",
    0: "Completed"
})
```

```python
import pandas as pd

# 1. Original file read pannrom
df = pd.read_csv("hotel_bookings.csv") # ungal original file name ithu thaan na?

# 2. Cleaning
# Missing values drop or fill
df = df.dropna(subset=['children', 'country']) # Example cleaning
df['children'] = df['children'].fillna(0)

# Duplicates remove
df = df.drop_duplicates()

# 3. New column
df['Stay_Nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

# 4. Save - MAIN STEP
df.to_csv("cleaned_hotel_bookings.csv", index=False)

print(f"✅ SUCCESS! {len(df)} rows saved to cleaned_hotel_bookings.csv")
print(df.head(3))
```
