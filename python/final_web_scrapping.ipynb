### Import Libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### Send Request To webpage

```python
url = "https://en.wikipedia.org/wiki/Tourism"

headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)

print(response.status_code)
```

### Create BeautifulSoup object

```python
soup = BeautifulSoup(
    response.text,
    "html.parser"
)
```

### Find webpage elements

```python
links = soup.find_all("a")
```

```python
len(links)
```

### Loop through elements

```python
tourism_data = []

for link in links:
    text = link.get_text(strip=True)

    if text:
        tourism_data.append(text)
```

```python
tourism_data[:20]
```

### Convert Scraped information to DataFrame

```python
tourism_df = pd.DataFrame(
    tourism_data,
    columns=["Tourism_Information"]
)
```

```python
tourism_df.head(10)
```

### Keep approximately 10 records

```python
tourism_df = tourism_df.head(10)
```

### Export csv

tourism_data.csv

```python
tourism_df.to_csv(
    "tourism_data.csv",
    index=False
)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://en.wikipedia.org/wiki/World_Tourism_rankings"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Correct table edukkrom
table = soup.find("table", {"class": "wikitable"})
rows = table.find_all("tr")

data = []
for row in rows[1:11]: # First 10 records
    cols = row.find_all(["th","td"])
    if len(cols) >= 2:
        country = cols[0].text.strip()
        # unwanted [1], \n ellam remove pannrom
        country = country.split('[')[0].strip()
        info = cols[1].text.strip().split('[')[0].strip()

        data.append({
            "Country": country,
            "Tourism Information": f"Most visited country, International tourist arrivals in millions: {info}"
        })

tourism_df = pd.DataFrame(data)
tourism_df = tourism_df.head(10)
print(tourism_df)

# Overwrite pannrom
tourism_df.to_csv("tourism_data.csv", index=False)
print("\n✅ Correct file saved!")
```
